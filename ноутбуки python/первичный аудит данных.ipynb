```python
#1
#Загружаем все данные в таблицу pandas.
df = pd.read_csv(
    file_path,                # Путь к исходному файлу.
    sep=";",                  # Разделитель — точка с запятой.
    encoding="utf-8",         # Кодировка, проверенная на образце.
    dtype=str,                # Все значения пока читаем как текст.
    keep_default_na=False    # Сохраняем исходные обозначения пропусков.
)

# Показываем количество строк и столбцов.
print("Размер таблицы:", df.shape)

# Показываем первые пять строк.
print("\nПервые пять строк:")
print(df.head())

# Показываем последние пять строк всего файла.
print("\nПоследние пять строк:")
print(df.tail())

# вычисляем использование оперативной памяти.
memory_bytes = df.memory_usage(deep=True).sum()

# Переводим байты в мегабайты (МиБ).
memory_mb = memory_bytes / (1024 ** 2)

# результат.
print("\nИспользование памяти:", round(memory_mb, 2), "МиБ")

# сравниваем названия столбцов с образцом.
print("\nСтолбцы совпадают:", df.columns.equals(sample.columns))

# Проверяем совпадение первых десяти записей.
print("Первые 10 строк совпадают:", df.head(10).equals(sample))

# Проверяем, что исходный файл по-прежнему существует.
print("Исходный файл существует:", file_path.is_file())

# Проверяем размер файла после чтения.
print("Размер исходного файла:", file_path.stat().st_size, "байт")
```

```python
#4
#первичная проверка пропусков (невозможность считывания символов)

# Выделяем только столбцы с электрическими измерениями (тисключаем из проверки дату и время).
measurement_columns = df.columns.drop(["Date", "Time"])

# Проверяем распространённые обозначения пропусков.
missing_markers = ["", "?", "NA", "N/A", "NaN", "null", "NULL"]

# Считаем каждое обозначение.
for marker in missing_markers:

    # сравнение каждой ячейки с каждым значением. На выходе true false таблица суммируется по индексам 0 и 1.
    counts = df[measurement_columns].eq(marker).sum()

    # общее число совпадений.
    total = counts.sum()

    # обозначение и количество дл каждого.
    print(repr(marker), ":", total)

# Проверяем пропуски, уже распознанные pandas как отсутсвующие.
print("\nРаспознанные pandas пропуски:")
print(df.isna().sum())

# сколько строк содержит хотя бы одно '?'.
rows_with_question = df[measurement_columns].eq("?").any(axis=1).sum()
print("\nСтрок с символом '?':", rows_with_question)
```

```python
#5 тк в предыдущем было много ? предполагаем что есть строки с ? и пустотой
#считаем пропуски по каждому столбцу,считаем их процент и ищем другие нечисловые
# Выбираем семь столбцов с измерениями
columns = df.columns.drop(["Date", "Time"])

# сравнение каждой ячейки с ?. На выходе true false таблица.
question_mask = df[columns].eq("?")

# сравнение каждой ячейки с пустой. На выходе true false таблица.
empty_mask = df[columns].eq("")

# Объединяем пропуски с помощью ИЛИ.
missing_mask = question_mask | empty_mask

# преобразования измерения в числа в отдельной таблице
# Неудачные преобразования превращаются в NaN в новой таблице numeric_data.
numeric_data = df[columns].apply(pd.to_numeric, errors="coerce")

# Находим нечисловые значения, отличающиеся от '?' и пустоты.
unknown_mask = numeric_data.isna() & ~missing_mask

# сводная таблица количества пропусков.
report = pd.DataFrame({
    "Символ ?": question_mask.sum(),
    "Пустые": empty_mask.sum(),
    "Другие": unknown_mask.sum()})
# Подсчитываем пропуски каждого столбца.
report["Всего"] = report.sum(axis=1)
# Вычисляем процент пропусков от количества строк.
report["Доля, %"] = (report["Всего"] / len(df) * 100).round(3)
print(report.to_string())

# число строк с хотя бы одним отсутствующим измерением.
print("\nСтрок с пропусками:",(missing_mask | unknown_mask).any(axis=1).sum())

# число строк, где отсутствуют все семь измерений.
print("Строк без всех семи измерений:", (missing_mask | unknown_mask).all(axis=1).sum())

# наличие других нечисловых обозначений.
print("Других нечисловых значений:", unknown_mask.sum().sum())

# Если обнаружены неизвестные значения, показываем примеры.
if unknown_mask.sum().sum() > 0:
    for column in columns:
        examples = df.loc[unknown_mask[column], column].unique()[:5]
        if len(examples) > 0:
            print(column, ":", examples)
#ожидания:большинство столбцов имеют одинаковую долю пропусков. если все строки содержат и ? и пустоты, то ожидания выполняются
```

```python
#6
#проверка есть ли отсутсвующие строки или строки с отсуствующим значением

# Объединяем дату и время в одну текстовую строку.
date_text = df["Date"].str.cat(df["Time"], sep=" ")

# Преобразуем текст в даты и время в виде специального типа datetime64 (2006-12-16 17:24:00).
timestamps = pd.to_datetime(
    date_text,                       # Текст с датой и временем.
    format="%d/%m/%Y %H:%M:%S",     # День, месяц, год и время.
    errors="coerce"                  # Ошибочные даты становятся NaT.
)

# даты, которые не удалось распознать.
print("Ошибок чтения дат:", timestamps.isna().sum())

# Показываем границы периода.
print("Начало измерений:", timestamps.min())
print("Конец измерений:", timestamps.max())

# Проверяем число временных меток.
print("Количество временных меток:", len(timestamps))

# Определяем разницу между соседними временными метками.
time_diff = timestamps.diff()

# Задаём ожидаемый шаг измерений — одну минуту.
one_minute = pd.Timedelta(minutes=1)

# правильные переходы длиной ровно одну минуту.
print("\nПереходов ровно по одной минуте:",
      (time_diff == one_minute).sum())

# переходы, превышающие одну минуту.
print("Переходов больше одной минуты:",
      (time_diff > one_minute).sum())

# Считаем повторения соседних временных меток.
print("Повторений подряд:",
      (time_diff == pd.Timedelta(0)).sum())

# Проверка не идёт ли время в обратном направлении.
print("Переходов назад во времени:",
      (time_diff < pd.Timedelta(0)).sum())

# Считаем повторные временные метки во всём файле.
print("Повторных временных меток:",
      (timestamps.notna() & timestamps.duplicated()).sum())

# Находим переходы с необычным временным шагом.
problem_mask = time_diff.notna() & (time_diff != one_minute)

# Показываем примеры проблемных переходов.
problems = pd.DataFrame({
    "Предыдущая метка": timestamps.shift(1),
    "Текущая метка": timestamps,
    "Разница": time_diff
})

# Выводим первые десять обнаруженных отклонений.
print("\nПримеры нарушений временного шага:")
print(problems.loc[problem_mask].head(10).to_string())

# Проверяем, что при преобразовании НЕ изменилось количество строк.
print("\nКоличество строк сохранилось:",
      len(timestamps) == len(df))

# Если есть ошибки чтения дат, показываем исходные значения.
if timestamps.isna().any():
    print("\nПримеры нераспознанных дат:")
    print(df.loc[timestamps.isna(), ["Date", "Time"]].head())
```

```python
#7 необязательный этап
#исследование цифрового отпечатка (неполная выгрузка данных)
from pathlib import Path
import hashlib

# Заново определяем путь к исходному файлу.
file_path = Path("/исходные_данные.txt")

# Проверяем наличие файла перед расчётом.
if not file_path.is_file():
    print("ОШИБКА: исходный файл не найден.")
else:
    # Создаём объект SHA-256.
    hash_object = hashlib.sha256()

    # Открываем файл только для чтения.
    with file_path.open("rb") as file:

        # Читаем небольшими блоками по 1 МиБ.
        for block in iter(lambda: file.read(1024 * 1024), b""):

            # Добавляем блок к расчёту контрольной суммы.
            hash_object.update(block)

    # Получаем цифровой отпечаток файла.
    file_hash = hash_object.hexdigest()

    # Выводим результаты проверки.
    print("SHA-256:", file_hash)
    print("Длина:", len(file_hash))
    print("Размер файла:", file_path.stat().st_size, "байт")
```
